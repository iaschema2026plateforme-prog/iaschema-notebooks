## Objectifs
À la fin de ce notebook, vous saurez charger un jeu de données, séparer les données, entraîner un modèle de classification et mesurer sa précision.

## 1. Importer les bibliothèques
Scikit-learn fournit les données, le modèle et les outils d’évaluation nécessaires.

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
print("Bibliothèques importées avec succès.")
````

## 2. Charger et examiner les données
Le jeu Iris contient 150 fleurs réparties en trois espèces. Chaque fleur est décrite par quatre mesures.

````python
iris = load_iris()
X = iris.data
y = iris.target
print("Dimensions :", X.shape)
print("Variables :", list(iris.feature_names))
print("Classes :", list(iris.target_names))
````

## 3. Créer les ensembles d’entraînement et de test
Le modèle apprend sur 80 % des données. Les 20 % restants servent à vérifier ses performances sur des exemples non vus.

````python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Entraînement :", len(X_train), "exemples")
print("Test :", len(X_test), "exemples")
````

## 4. Entraîner le modèle
Nous utilisons un arbre de décision limité à trois niveaux afin de conserver un modèle simple.

````python
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)
print("Modèle entraîné.")
````

## 5. Évaluer les prédictions
La précision représente la proportion de prédictions correctes.

````python
predictions = model.predict(X_test)
score = accuracy_score(y_test, predictions)
print(f"Précision : {score:.2%}\n")
print(classification_report(y_test, predictions, target_names=iris.target_names))
````

## 6. Faire une nouvelle prédiction
L’exemple ci-dessous représente quatre mesures d’une fleur en centimètres.

````python
nouvelle_fleur = [[5.1, 3.5, 1.4, 0.2]]
classe = model.predict(nouvelle_fleur)[0]
print("Espèce prédite :", iris.target_names[classe])
````
