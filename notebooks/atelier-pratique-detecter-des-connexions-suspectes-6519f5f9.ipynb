## 1. Charger les données

Le jeu de données ci-dessous simule quelques tentatives de connexion à une application d'entreprise.

```python
import pandas as pd

donnees = {
    'utilisateur': ['amina', 'paul', 'amina', 'sara', 'paul', 'paul', 'yassine', 'amina'],
    'adresse_ip': ['192.168.1.10', '10.0.0.8', '203.0.113.25', '192.168.1.15',
                   '10.0.0.8', '198.51.100.42', '192.168.1.20', '203.0.113.25'],
    'statut': ['Succès', 'Échec', 'Échec', 'Succès', 'Échec', 'Échec', 'Succès', 'Échec']
}

df = pd.DataFrame(donnees)
df
```

## 2. Afficher uniquement les connexions échouées

```python
echecs = df[df['statut'] == 'Échec']
echecs
```

## 3. Compter les échecs par utilisateur

Plusieurs échecs rapprochés peuvent justifier une vérification.

```python
resume = echecs.groupby('utilisateur').size().sort_values(ascending=False)
resume
```

## 4. Identifier les comptes à surveiller

```python
seuil = 2
comptes_a_verifier = resume[resume >= seuil]
print('Comptes avec au moins', seuil, 'échecs :')
print(comptes_a_verifier)
```
